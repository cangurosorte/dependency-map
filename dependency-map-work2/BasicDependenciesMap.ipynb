```python
import pandas as pd
from jaal import Jaal
from jaal.datasets import load_got
```

```python

sz_file_path = "csv/DependenciasBRCLbase.csv"
sz_file_csv = pd.read_csv(sz_file_path,usecols=['initiatingDeviceName','receivingDeviceName','receivingPort','connectionCount'],
                          header=0,
                        
                        )
```

```python
df = sz_file_csv.rename(
    columns={
        "initiatingDeviceName": "from",
        "receivingDeviceName": "to",
        "connectionCount": "weight"
    }
)
```

```python
df_grouped = df.groupby(['from','to'], as_index=False)['weight'].sum()
```

```python

Jaal(df_grouped).plot(port=8075,vis_opts={'height': '1200px', # change height
                                     'interaction':{'hover': True}, # turn on-off the hover 
                                   # 'physics':{'stabilization':{'iterations': 100}}})   # Con Fisica
                                    'physics':{'enabled': False}})                     # Estatico





#
# Abrir en http://127.0.0.1:8050/
# Links de los Datos https://visjs.github.io/vis-network/docs/network/physics.html#
# https://github.com/imohitmayank/jaal
```
